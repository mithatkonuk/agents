# And now - TOOLS!

Let's start with a function...

````python
def record_email_tool(email):
    print(f"Tool called to record an email: {email}")
    with open("emails.txt", "a", encoding="utf-8") as f:
        f.write(email + "\n")
    return "Email received"
````

````python
record_email_tool("test@testy.com")
````

## Step 1 - write some json to describe the tool


````python
record_email_tool_json = {
    "name": "record_email_tool",
    "description": "Use this tool to record that a user provided their email address",
    "parameters": {
        "type": "object",
        "properties": {
            "email": {"type": "string", "description": "The email address of this user"}
        },
        "required": ["email"],
        "additionalProperties": False
    }
}
````

````python
tools = [{"type": "function", "function": record_email_tool_json}]
````

````python
tools
````

## Step 2 - a new chat() function

This is where we implement the tool call.

The reality is, it's a bit clunky. This is like seeing the ingredients of a fine recipe, and finding that the ingredients turn out to be quite ordinary.

Tool calling is an "if" statement. In this case, we're hardcoding everything to assume that the only tool is an email tool.

SIDENOTE: If you're thinking - but wait! I should be remembering this so I can do it myself! Then the key point is: this is what Agent Frameworks take care of for you. In practice, you'll likely never type this again yourself. We are shielded from these if statements by the Agent Framework. That's why they're often described as "abstraction layers".

````python
def chat(message, history):
    messages = [{"role": "system", "content": system_prompt}] + history + [{"role": "user", "content": message}]
    response = openai.chat.completions.create(model="openrouter/free", messages=messages, tools=tools)
    print(response.model_dump_json(indent=2))
    if response.choices[0].finish_reason=="tool_calls":
            message = response.choices[0].message
            tool_call = message.tool_calls[0]
            email = json.loads(tool_call.function.arguments).get("email")
            record_email_tool(email)
            messages.append(message)
            messages.append({"role": "tool", "content": "Email recorded", "tool_call_id": tool_call.id})
            response = openai.chat.completions.create(model="openrouter/free", messages=messages, tools=tools)
            
    return response.choices[0].message.content
````

````python
gr.ChatInterface(chat).launch(inbrowser=True)
````

## Step 3

Our first ever Agent Loop, done without an Agent Framework!

Changes:
1. Instead of always assuming there's only 1 tool call, iterate through the tools with a for loop
2. Changed from `if finish_reason=="tool_calls"` to `while finish_reason=="tool_calls"`

````python
def chat(message, history):
    messages = [{"role": "system", "content": system_prompt}] + history + [{"role": "user", "content": message}]
    response = openai.chat.completions.create(model="openrouter/free", messages=messages, tools=tools)
         
    while response.choices[0].finish_reason=="tool_calls":
            message = response.choices[0].message
            messages.append(message)
            for tool_call in message.tool_calls:
                email = json.loads(tool_call.function.arguments).get("email")
                record_email_tool(email)
                messages.append({"role": "tool", "content": "Email recorded", "tool_call_id": tool_call.id})
            response = openai.chat.completions.create(model="openrouter/free", messages=messages, tools=tools)
            
    return response.choices[0].message.content
````

````python
gr.ChatInterface(chat).launch(inbrowser=True)
````

````python
#### Different Project - Restaurant Assistant


from dotenv import load_dotenv
from openai import OpenAI
from pypdf import PdfReader
from IPython.display import Markdown, display
import gradio as gr
import os
import json

load_dotenv(override=True)
api_key = os.getenv("OPENROUTER_API_KEY")
base_url = os.getenv("OPENROUTER_BASE_URL")
openai = OpenAI(api_key=api_key, base_url=base_url)



summary = f"""
"Lumen is a modern German-Mediterranean bistro in Berlin-Mitte, founded in 2021 by chef Anna Berger. 
 Known for seasonal small plates, a natural wine list, and a relaxed but refined atmosphere. Mid-range pricing (€€€)."
"""

menu = f"""
Starters: Burrata with heirloom tomato (contains dairy) – €12 ... Mains: Lamb shoulder, harissa jus (contains none of the top 14 allergens) – €26 ...
"""
chef_bio = f"""
Chef Anna Berger trained at Tantris Munich and worked under [X] in Copenhagen before opening Lumen. Her philosophy centers on hyper-seasonal, zero-waste cooking.
"""
policies = f"""
Open Tue–Sun, 18:00–23:00. Reservations recommended, held for 15 min. No dress code. Wheelchair accessible. Cancellations under 24h may incur a €20/person fee for parties of 6+.
"""


system_prompt = f"""

# Your role

You are a digital concierge running on a restaurant's website, chatting with visitors who are curious about the restaurant.
You represent the restaurant and speak on its behalf, as if you were the maître d' greeting a guest.
You answer questions related to the menu, ingredients, allergens, reservations, hours, location, and the dining experience.

Here are the details of the restaurant you are representing:

{summary}

If asked, you explain clearly that you are an AI concierge built for this restaurant, not a human staff member.

# Context

Here is the current menu, chef's background, and house policies so that you can answer questions accurately:

{menu}

{chef_bio}

{policies}

# Rules

Engage with the visitor. Be warm, hospitable, and precise, as if speaking to a guest deciding whether to book a table or place an order.
Avoid answering questions that are not related to the menu, ingredients, dining experience, reservations, or the restaurant itself;
steer the conversation back to these topics.

Always stay in character as the restaurant's concierge. Represent the restaurant, not yourself as a general assistant.

IMPORTANT: Never guess about ingredients, allergens, or availability. If you don't know the answer — especially about allergens, substitutions, or off-menu requests — say so clearly and suggest the guest confirm with staff before ordering. Never make up an answer.
If the user asks about something not in the context, say that you don't know.
"""


business_supervisor_prompt = f"""
You are a business supervisor. You will check answer from the LLM and be realistic, the answer should not out of the context of the business. if it is not related to the business, you should reject it.
If it is related to the business, return "yes"
If it is not related to the business, return "no"

here business context is the summary, menu, chef_bio, and policies of the restaurant

{system_prompt}
"""



check_availability_tool_json = {
    "name": "check_availability_tool",
    "description": "Check table availability for a given date, time, and party size",
    "parameters": {
        "type": "object",
        "properties": {
            "date": {"type": "string", "description": "Reservation date, format YYYY-MM-DD"},
            "time": {"type": "string", "description": "Reservation time, format HH:MM (24h)"},
            "party_size": {"type": "integer", "description": "Number of guests"}
        },
        "required": ["date", "time", "party_size"],
        "additionalProperties": False
    }
}



record_reservation_tool_json = {
    "name": "record_reservation_tool",
    "description": "Record a confirmed table reservation for a guest",
    "parameters": {
        "type": "object",
        "properties": {
            "name": {"type": "string", "description": "Guest's name"},
            "email": {"type": "string", "description": "Guest's email for confirmation"},
            "date": {"type": "string"},
            "time": {"type": "string"},
            "party_size": {"type": "integer"},
            "special_requests": {"type": "string", "description": "Allergies, seating preference, occasion, etc."}
        },
        "required": ["name", "email", "date", "time", "party_size"],
        "additionalProperties": False
    }
}


def check_availability_tool(date, time, party_size):
    return {"available": True, "date": date, "time": time, "party_size": party_size}

def record_reservation_tool(name, email, date, time, party_size, special_requests):
    return {"status": "confirmed", "name": name, "date": date, "time": time, "party_size": party_size}
    

### this is the business keeper, check answer from the LLM to evaluate if it is related to the business or reject it 
business_supervisor = openai.chat.completions.create(model="openrouter/free", messages=[{"role": "system", "content": system_prompt}])

tools = [
    {"type": "function", "function": check_availability_tool_json},
    {"type": "function", "function": record_reservation_tool_json},
]

def chat(user_message, history):
    messages = [{"role": "system", "content": system_prompt}] + history + [{"role": "user", "content": user_message}]
    response = openai.chat.completions.create(model="openrouter/free", messages=messages, tools=tools)

    while response.choices[0].finish_reason == "tool_calls":
        message = response.choices[0].message
        messages.append(message)

        for tool_call in message.tool_calls:
            args = json.loads(tool_call.function.arguments)

            if tool_call.function.name == "check_availability_tool":
                result = check_availability_tool(
                    args.get("date"), args.get("time"), args.get("party_size")
                )
            elif tool_call.function.name == "record_reservation_tool":
                result = record_reservation_tool(
                    args.get("name"), args.get("email"), args.get("date"),
                    args.get("time"), args.get("party_size"), args.get("special_requests")
                )
            else:
                result = {"error": f"Unknown tool {tool_call.function.name}"}

            messages.append({
                "role": "tool",
                "content": json.dumps(result),
                "tool_call_id": tool_call.id
            })

        response = openai.chat.completions.create(model="openrouter/free", messages=messages, tools=tools)

    final_content = response.choices[0].message.content

    # supervisor check
    check = openai.chat.completions.create(
        model="openrouter/free",
        messages=[
            {"role": "system", "content": business_supervisor_prompt},
            {"role": "user", "content": f"Guest asked: {user_message}\nConcierge answered: {final_content}\nIs this on-topic? Answer yes or no."}
        ]
    ).choices[0].message.content.strip().lower()

    if "no" in check:
        return "I'm here to help with questions about Lumen — our menu, reservations, or dining experience. Is there something along those lines I can help with?"
    return final_content


gr.ChatInterface(chat).launch(inbrowser=True, server_port=7861)


````
